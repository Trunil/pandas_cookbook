```python
import numpy as np
import pandas as pd
```

Create dataframe with sample names and gender. Gender should be a categorical column.

```python
# dataframee with sample names and gender
df = pd.DataFrame({
    'Sample' : ['Sample_A', 'Sample_B', 'Sample_C', 'Sample_D', 'Sample_E'],
    'Gender' : ['Male', 'Female', 'Female', 'Male', 'Female']
})

df
```

See the data type of the columns.

df.info()

Both columns are of `object` dtype. Convert `Gender` column to categorical.

```python
df = df.astype({'Gender' : 'category'})
df.info()
```

Convert to ordered categories

```python
from pandas.api.types import CategoricalDtype

categories = CategoricalDtype(categories=['Female', 'Male'], ordered=True)

df = df.astype({'Gender' : categories})
df.info()
```
