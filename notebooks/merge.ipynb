```python
import numpy as np
import pandas as pd
```

```python
# first dataframe
df1 = pd.DataFrame({
    'id': ['a', 'b', 'c', 'd'],
    'prop_1' : [4,8,9,2],
    'prop_2' : [1,6,5,2]
})


# second dataframe
df2 = pd.DataFrame({
    'id': ['g', 'd', 'h', 'a'],
    'prop_1' : [8,5,0,3],
    'prop_2' : [5,1,7,4]
})
```

```python
df1
```

```python
df2
```

```python
# merge with common values in `id` column
df3 = pd.merge(df1, df2, on='id', how='inner')
df3
```

```python
# merge with values in `id` column of df1 only
df3 = pd.merge(df1, df2, on='id', how='left')
df3
```

```python
# merge with values in `id` column of df2 only
df3 = pd.merge(df1, df2, on='id', how='right')
df3
```

```python
# add suffixes
# merge with common values in `id` column
df3 = pd.merge(df1, df2, on='id', how='inner', suffixes=['_df1', '_df2'])
df3
```
