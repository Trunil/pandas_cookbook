```python
import numpy as np
import pandas as pd
```

```python
# create dataframe
df = pd.DataFrame({
    'name' : ['A', 'B', 'A', 'B','A', 'B', 'A', 'B'],
    'col_a' : np.random.randint(1,20, size=8),
    'col_b' : np.random.randint(1,20, size=8),
})
df
```

```python
# groupby categories in `name`
# take sum of values for that category
df.groupby('name')[['col_a', 'col_b']].sum()
```

```python
# groupby categories in `name`
# take mean of values for that category
df.groupby('name')[['col_a', 'col_b']].mean()
```
